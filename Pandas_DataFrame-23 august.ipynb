<div style="background-color: #ADD8E6; padding: 12px; border-radius: 5px; border-left: 5px solid #0066cc;">
  <h3>2. Data Cleansing</h3>
  
</div>


```python
orders = pd.DataFrame({
    'Order ID':        [1001, 1002, 1003, 1004, 1005, 1005, 1006, 1007],
    'Customer Name':   ['Liam Ortiz', 'Nora Chen', np.nan, 'Sara Malik', 'Tom Reyes', 'Tom Reyes', 'Uma Iyer', np.nan],
    '  Order Amount':  [250.0, np.nan, 180.5, 90.0, 310.75, 310.75, np.nan, 145.0],
    'ship_date ':      ['2024-01-05', '2024-01-06', '2024-01-07', np.nan, '2024-01-09', '2024-01-09', '2024-01-10', '2024-01-11'],
    'Region':          ['North', 'South', 'South', 'East', 'West', 'West', np.nan, 'North']
})

orders
```

### 2.1 Standardize the columns 

```python
#1. Strip the leading/trailing whitespace from every column name. 

orders.columns = orders.columns.str.strip()

orders
```

```python
#2. Convert all column names to upper case letters

orders.columns = orders.columns.str.capitalize()

orders
```

```python
#3. Add underscore between two words

orders.columns = orders.columns.str.replace(' ','_')

orders.columns
```

```python
#4. After standardizing names in Q7/Q8, rename the column `order_id` to `order_number`.

orders.rename(columns = {'Order_id': 'order_number'},inplace=True)

orders.columns
```

### 2.2 Working with NULL values

```python
#1. Check how many missing values exist in each column.

orders.isna().sum()
```

```python
#2 Find the rows where customer name is missing

orders[orders['Customer_name'].isna()]
```

```python
#3. Drop the rows where `Customer Name` is missing.

orders.dropna(subset=['Customer_name'],axis=0)
```

```python
# 4. Fill missing values in `Region` with the string `Unknown`.

orders['Region'].fillna(value='Unknown')
```

### 2.3 Working with duplicate values

```python
#1. Show Summary of duplicate rows

orders.duplicated().sum()
```

```python
#2. Show the duplicate rows
orders[orders.duplicated(keep=False)]
```

```python
#3. Drop duplicate rows, keeping the first occurrence.

orders.drop_duplicates(keep='first',inplace=True)
```

```python
#3. Fill missing values in `Order Amount` with that column's mean.

#3.1 Find the mean

meanval = orders['Order_amount'].mean()
```

```python
#3.2 Impute the mean

orders['Order_amount'].fillna(value=meanval)
```

```python
#5. Reset the DataFrame's index so it runs cleanly from 0 after all the cleaning above.
orders.reset_index(inplace=True)
```
